# **1.1 Overall Sales Trends:**


1. How have total sales varied across quarters?
2.  Which quarter generates the highest/lowest sales?

To analyse our standing in terms of sales in the market each quarter of 2020 & 2021 and to find out which quarter/season is underperforming and overperforming.

```python
#1. How have total sales varied across quarters?

# We ensure that 'Invoice Date' is in datetime format,either way we need to change it
data_sales['Invoice Date'] = pd.to_datetime(data_sales['Invoice Date'])

# Most importantly, we need to extract year and quarter from 'Invoice Date'
data_sales['Year'] = data_sales['Invoice Date'].dt.year
data_sales['Quarter'] = data_sales['Invoice Date'].dt.quarter

# Afterwards we should filter data for the years 2020 and 2021
filtered_data = data_sales[(data_sales['Year'] == 2020) | (data_sales['Year'] == 2021)]

# Following the filtration of data for years 2020-2021 we need to add the start and end dates of each quarter
filtered_data['Quarter_Start_Date'] = filtered_data['Invoice Date'].dt.to_period('Q').apply(lambda x: x.start_time)
filtered_data['Quarter_End_Date'] = filtered_data['Invoice Date'].dt.to_period('Q').apply(lambda x: x.end_time)

# Therefor, we need to group by 'Year' and 'Quarter' to calculate total sales
quarterly_sales = filtered_data.groupby(['Year', 'Quarter']).agg(
    Total_Sales=('Total Sales', 'sum'),

).reset_index()

# Since, given output was not well readible, we are adding to 'Total Sales' a dollar sign and commas for better understanding
quarterly_sales['Total_Sales'] = quarterly_sales['Total_Sales'].apply(lambda x: f"${x:,.2f}")

# To sum it up the total sales for all quarters for 2020 and 2021
print("Total Sales for Each Quarter (2020 and 2021):")
print(quarterly_sales)
```

From output we can see total sales per quarter, but for more clarity we can look at vizualisation in which we can surely see differences berween quarters.For vizualisation we are using bar chart, since A bar chart because it clearly displays quarterly sales trends, highlights high/low sales periods, and makes it easy to compare different years and quarters for decision-making.

```python
#2.  Which quarter generates the highest/lowest sales?

# As previously we had to ensure that the 'Invoice Date' is in datetime format
data_sales['Invoice Date'] = pd.to_datetime(data_sales['Invoice Date'])

# To simulate accurate interpretation of the highest and lowest sales, we need to extract year and quarter from 'Invoice Date'
data_sales['Year'] = data_sales['Invoice Date'].dt.year
data_sales['Quarter'] = data_sales['Invoice Date'].dt.quarter

# To emphasize better understanding of the yearly changes, we need to filter for the years 2020 and 2021
filtered_data = data_sales[(data_sales['Year'] == 2020) | (data_sales['Year'] == 2021)]

# Using the ".apply()" and ".to_period('Q')" we are adding the start and end dates of each quarter
filtered_data['Quarter_Start_Date'] = filtered_data['Invoice Date'].dt.to_period('Q').apply(lambda x: x.start_time)
filtered_data['Quarter_End_Date'] = filtered_data['Invoice Date'].dt.to_period('Q').apply(lambda x: x.end_time)

# To sum the data into one chart we are grouping the data by 'Year' and 'Quarter' to calculate total sales
quarterly_sales = filtered_data.groupby(['Year', 'Quarter']).agg(
    Total_Sales=('Total Sales', 'sum'),
    Start_Date=('Quarter_Start_Date', 'first'),
    End_Date=('Quarter_End_Date', 'first')
).reset_index()

# For Clarity of quarterly growth we create labels for new column
quarterly_sales['Quarter_Label'] = quarterly_sales['Year'].astype(str) + "-Quarter" + quarterly_sales['Quarter'].astype(str)

# To format quarterly sales we need to plot it using the 'Total Sales'
quarterly_sales['Total_Sales_Value'] = quarterly_sales['Total_Sales']

# We are plotting the bar chart.
plt.figure(figsize=(12, 6))
sns.barplot(x='Quarter_Label', y='Total_Sales_Value', data=quarterly_sales, palette='Blues_d')
plt.title("Quarterly Sales (2020 and 2021)", fontsize=16)
plt.xlabel("Quarter", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

**Sales Trend for 2020:**

- The overall sales for the first three quarters of 2020 were consistent, ranging from 48 million to 55 million dollars.
- However, Q4 2020 saw a sharp decline to $27M, which may have been caused by outside circumstances (e.g., supply chain concerns, pandemic impact).

# **1.2	Regional Performance:**


1.  What is the sales contribution of each region to total revenue?
2.  Vizualization to see which region has the highest sales?

We are doing this analysis to classify which regions have been performing the best and which regions need attention from the Adidas Sales Team, to improve thier standing in the regional market. It will help us in finding loopholes within the performance executed in the country and will make our task of monitoring more effecient.


```python
#1.  What is the sales contribution of each region to total revenue?

# To Classify the regional data, we need to group by 'Region' to calculate total sales
region_sales = data_sales.groupby('Region').agg(
    Total_Sales=('Total Sales', 'sum')
).reset_index()

# Then its is important to calculate total revenue across all regions
total_revenue = region_sales['Total_Sales'].sum()

# To analyze effeciently we need to calculate the sales contribution percentage for each region
region_sales['Sales_Percentage'] = (region_sales['Total_Sales'] / total_revenue) * 100

# Sorting of the data by 'Total_Sales' in descending order will help us in understanding from figures from highest to lowest
region_sales = region_sales.sort_values(by='Total_Sales', ascending=False)

# Formatting of the Total Sales and Sales Percentage for better understanding
region_sales['Total_Sales'] = region_sales['Total_Sales'].apply(lambda x: f"${x:,.2f}")
region_sales['Sales_Percentage'] = region_sales['Sales_Percentage'].apply(lambda x: f"{x:.2f}%")

# Now we can print the result
print("Sales Contribution Per Region (Percentage of Total Sales):")
print(region_sales)

```

```python
# 2.  Vizualization to see which region has the highest sales?

# Let us prepare a heatmap for visualising region's higest sales
region_sales['Total_Sales_Value'] = region_sales['Total_Sales'].str.replace('[\$,]', '', regex=True).astype(float)
heatmap_data = region_sales[['Region', 'Total_Sales_Value']].set_index('Region')

# Then we can put the factors to plot a graph for the visualization
plt.figure(figsize=(8, 6))
sns.heatmap(
    heatmap_data,
    annot=heatmap_data.applymap(lambda x: f"${x:,.2f}"),
    fmt="",
    cmap="Greens",
    cbar=True
)
plt.title("Sales Intensity by Region", fontsize=16)
plt.ylabel("Region", fontsize=12)
plt.xlabel("Total Sales ($)", fontsize=12)
plt.show()


```

**Details:**
- With roughly 269.9 million in total sales, the West is the best-performing region, according to the heatmap. The Northeast comes in second with 186.3 million and the Southeast with 163.1 million.
- On the flip side, the Midwest (135.8M) and South (144.6M) perform the worst, suggesting areas that could need additional focus.

- In order to maximize sales strategies develop distribution, and improve overall market performance, the Adidas Sales Team uses this information to determine which regions are strong and weak.
- For Improved revenue growth and market penetration the Adidas Team needs to focus on concentrating on underperforming areas.

# **1.3	City-Level Insights:**

1.  Which cities drive the most sales?

We need to understand how different cities and states been performing in the last 2 years, because some cities that have been underperforming can be taken under due dilligence. This would help us distribute our reseources based on the returns and market demand for our product.


```python
#1.  Which cities drive the most sales?

# To calculate the highest sales first we need to group by 'State' to calculate total sales, sales count, and operating profit
state_sales = data_sales.groupby('State').agg(
    Total_Sales=('Total Sales', 'sum'),
    Sales_Count=('Total Sales', 'count'),
    Total_Operating_Profit=('Operating Profit', 'sum')
).reset_index()

# Sorting of the data is important so that we could reorder it in the descending order
state_sales = state_sales.sort_values(by='Total_Sales', ascending=False)

# For Better Understanding we need to format the 'Total Sales' and 'Total Operating Profit' columns
state_sales['Total_Sales'] = state_sales['Total_Sales'].apply(lambda x: f"${x:,.0f}")
state_sales['Total_Operating_Profit'] = state_sales['Total_Operating_Profit'].apply(lambda x: f"${x:,.0f}")

# Then we need to print state-wise results
print("State-wise Total Sales and Operating Profit Analysis (Descending Order):")
print(state_sales)
```

- With roughly 64.2 million in total sales and the highest operating profit of 23.3 million, New York is the state with the best performance. Florida (59.2 million) and California (60.1 million) are next in line, both displaying high profitability and sales. Despite having a little lower profit margin, Texas (46.3 million) is equally ranked well.

- States with the lowest sales and operating profits on the lower end, such as Iowa (7.4 million), Minnesota (7.3 million), and Nebraska (5.9 million), indicate potential locations for market growth. The Adidas Sales Team should concentrate on states that aren't performing well in order to improve their sales tactics and expand their market share.


```python
#2.  Are there cities with have High, Medium & Low Sales1

# This code calculates the total sales for each city,classifies cities into low,medium and high sales categories based on their total sales.
# To Calculate the city-wise total sales
total_sales_by_city = data_sales.groupby('City').agg(
    Total_Sales=('Total Sales', 'sum')
).reset_index()

def classify_city_sales(row):
    if row['Total_Sales'] < 10_000_000:  # Less than $10 million
        return 'Low Sales'
    elif 10_000_000 <= row['Total_Sales'] < 20_000_000:  # Between $10M and $20M
        return 'Medium Sales'
    else:  # $20 million or above
        return 'High Sales'

# To Classify cities into Sales Cartegories, we need to apply the function
total_sales_by_city['Sales Category'] = total_sales_by_city.apply(classify_city_sales, axis=1)


# We need to symbolise the Total Sales using the $ symbol and commas
total_sales_by_city['Total_Sales'] = total_sales_by_city['Total_Sales'].apply(lambda x: f"${x:,.0f}")

# Print the updated dataframe based on the categories
print("Cities Classified into Sales Categories:")
print(total_sales_by_city[['City', 'Total_Sales', 'Sales Category']])

```

**Cities with High Sales:**

Cities with the highest total sales, such as New York (39.8M), Charleston (39.9M), and San Francisco (34.5M), are important sources of income for Adidas.
Other top-performing cities with steady high sales include Miami, Los Angeles, Orlando, and Seattle.

**Cities with Medium Sales:**

Cities that do well but fall outside of the top sales level include Dallas, Denver, Honolulu, and Charlotte.
With the correct sales and marketing tactics, these cities have room for development.

**Cities with Low Sales:**

The cities with the lowest overall sales are Sioux Falls (8.4M), Omaha (5.9M), and Fargo (7.7M).
Adidas needs to assess the state of the market and look into ways to boost sales in these sectors.


```python
#3.  Sales Comparison between 2020 & 2021

# For Clarity to the Invoice Date we need to check if it is in Datetime Format
data_sales['Invoice Date'] = pd.to_datetime(data_sales['Invoice Date'])

# For Yearly analysis, we need to extract Year from 'Invoice Date'
data_sales['Year'] = data_sales['Invoice Date'].dt.year

# Based on the chosen dataset we need to filter data for 2020 and 2021
filtered_sales = data_sales[(data_sales['Year'] == 2020) | (data_sales['Year'] == 2021)]

# Grouping of the Total Sales per year based on 'City' and 'Year'
city_sales_by_year = filtered_sales.groupby(['City', 'Year']).agg(
    Total_Sales=('Total Sales', 'sum')
).reset_index()

# After Grouping, Pivot the chosen data to make 'Year' columns (2020 and 2021)
sales_diff = city_sales_by_year.pivot(index='City', columns='Year', values='Total_Sales').reset_index()

# For understanding the diffrentiated columns we need to rename them
sales_diff.columns = ['City', 'Sales_2020', 'Sales_2021']

# Fill NaN values with 0 (if any city has sales in only one year)
sales_diff.fillna(0, inplace=True)

# For 2020 & 2021 we need to make the columns readable and clear
sales_diff['Sales_2020'] = sales_diff['Sales_2020'].apply(lambda x: f"${x:,.0f}")
sales_diff['Sales_2021'] = sales_diff['Sales_2021'].apply(lambda x: f"${x:,.0f}")

# Finally, Printing the table
print("City-wise Sales for 2020 and 2021 with Percentage Change:")
print(sales_diff)
```

**Conclusion: City-wise Sales Analysis for 2020 and 2021**

According to the data, city-wise sales changed significantly between 2020 and 2021. While some areas saw large increases in sales, others had no sales in 2021 despite having strong sales in 2020.

Major cities with high sales in 2020, including Denver (21 million), Houston (25.4 million), Las Vegas (20.8 million), New York (39.8 million), Miami (31.6 million), and Seattle (26.3 million), did not have any sales in 2021.

**The following cities had sales in both years:**

We need to outline that in outcome we can see that sales in Philadelphia increased from 948K in 2020 to 9.4M in 2021,As well as overal sales in San Francisco drastically rebusted from 8.7 million in 2020 to 25.8 million in 2021.


```python
# 4.  Which States are top performing in the USA (>50 Million)

# For the analysis we need to analyse the data of Total Sales in the State-Wise manner
state_sales = data_sales.groupby('State').agg(
    Total_Sales=('Total Sales', 'sum')
).reset_index()

# Renaming of the columns to 'State Name' and 'Total Sales'
state_sales.columns = ['State Name', 'Total Sales']

# To filter the Total Sales which is higher than 50 Million we need to use subsetting
high_sales_states = state_sales[state_sales['Total Sales'] > 50_000_000]

# Total sales is sorted in the descending order and later Top 3 are chosen
top_3_sales_states = high_sales_states.sort_values(by='Total Sales', ascending=False).head(3)

# We add commas to Total Sales Values and also the $ symbol
top_3_sales_states['Total Sales'] = top_3_sales_states['Total Sales'].apply(lambda x: f"${x:,.0f}")

# Top 3 states are then later subsetted to include only 'State Name' and 'Total Sales'
subset_top_3_sales = top_3_sales_states[['State Name', 'Total Sales']]

# Print the data
print("Subset of Rows (States with Total Sales Above 50 Million):")
print(subset_top_3_sales)
```

```python
#1. Which products generate the most revenue?

# First to calculate the Total Units sold we need to group by 'City', 'Product', and 'Gender'
city_product_units = data_sales.groupby(['City', 'Product', 'Gender']).agg(
    Total_Units_Sold=('Units Sold', 'sum')
).reset_index()

# With this command we are able to find the best-performing product for each city
best_product_per_city = city_product_units.loc[
    city_product_units.groupby('City')['Total_Units_Sold'].idxmax()
]

# Total Units Sold is sorted in descending order for better readability and overall understanding
best_product_per_city = best_product_per_city.sort_values(by='Total_Units_Sold', ascending=False)

# Display the results
print("Best-Performing Product for Each City with Gender Information:")
print(best_product_per_city)

```

# **Hypothetical Business Scenarios**
1.	If online sales were increased by 10%, what would be the impact on overall revenue?
2.	What happens if a particular retailer (e.g., the top performer) leaves a key region?

We need to analyze hypothetical scenarios for company that can better prepare company for growth and reduce risks by analyzing these scenarios. Potential income gains can be seen by a 10% rise in online sales, which directs digital investments and marketing tactics. Adidas can plan for potential income losses by assessing the effects of a major retailers leaving an important region. This enables them to strengthen their relationships, increase direct-to-consumer sales, or gain new potential partnerships in other locations. These observations guarantee Adidas maintains its competitiveness and successfully adjusts to shifts in the market.

```python
# 1.	If online sales were increased by 10%, what would be the impact on overall revenue?

from sqlalchemy import text

# Step 1: As for first step we are query total revenue and total online sales
query = text("""
    SELECT
        SUM(Total_Sales) AS Total_Revenue,
        (SELECT SUM(Total_Sales) FROM sales_data WHERE Sales_Method = 'Online') AS Online_Sales
    FROM sales_data
""")

with engine.connect() as connection:
    result = connection.execute(query)
    row = result.fetchone()  # We need to fetch the first row
    total_revenue = row[0]  # This will access the first column (Total_Revenue)
    online_sales = row[1]  # This will access the second column (Online_Sales)

# Step 2: As the second step we need to calculate the impact of a 10% increase in online sales
increased_online_sales = online_sales * 1.10  # With this command we are specifying to increase online sales by 10%
new_total_revenue = (total_revenue - online_sales) + increased_online_sales  # Afterwards we need to adjust total revenue

# Step 3: With this step we are calculating the change in revenue
revenue_difference = new_total_revenue - total_revenue

# Finally we need to print the results
print(f"Current Total Revenue: ${total_revenue:,.2f}")
print(f"Current Online Sales: ${online_sales:,.2f}")
print(f"New Total Revenue (with 10% increase in online sales): ${new_total_revenue:,.2f}")
print(f"Impact on Overall Revenue: ${revenue_difference:,.2f}")
```

### **Conclusion: Impact of a 10% Increase in Online Sales on Total Revenue**  

This analysis evaluates the effect of a 10% increase in online sales on overall revenue, that will shift from 899.9M USD to 924.6M, which will cause overall revenue incresing to 24.7M. Focusing business strategy on online growth by investing in digital marketing,promotions and e-commerce optimization,online exlusive deals or offerings can help achieving this goals.


```python
# 2.	What happens if a particular retailer (e.g., the top performer) leaves a key region?

from sqlalchemy import text

# Step 1: First of all we identifying the top-performing retailer in the key region
region = "Southeast"  # Specifying our key region, that as we saw previously is southeast
query_top_retailer = text(f"""
    SELECT
        Retailer,
        SUM(Total_Sales) AS Retailer_Sales
    FROM sales_data
    WHERE Region = '{region}'
    GROUP BY Retailer
    ORDER BY Retailer_Sales DESC
    LIMIT 1
""")

with engine.connect() as connection:
    result = connection.execute(query_top_retailer)
    top_retailer_data = result.fetchone()
    top_retailer = top_retailer_data[0]
    top_retailer_sales = top_retailer_data[1]

# Step 2: Afterwards we need to calculate the region's total sales
query_region_sales = text(f"""
    SELECT
        SUM(Total_Sales) AS Region_Sales
    FROM sales_data
    WHERE Region = '{region}'
""")

with engine.connect() as connection:
    result = connection.execute(query_region_sales)
    region_sales = result.fetchone()[0]

# Step 3: To receive our results we need to calculate the sales without the top-performing retailer
sales_without_top_retailer = region_sales - top_retailer_sales

# Step 4:Now we need to calculate the impact
impact_percentage = ((region_sales - sales_without_top_retailer) / region_sales) * 100

# Finally we need to print the results
print(f"Region: {region}")
print(f"Top-Performing Retailer: {top_retailer}")
print(f"Sales by {top_retailer}: ${top_retailer_sales:,.2f}")
print(f"Total Region Sales: ${region_sales:,.2f}")
print(f"Region Sales Without {top_retailer}: ${sales_without_top_retailer:,.2f}")
print(f"Impact on Region Sales: {impact_percentage:.2f}%")
```

### **Conclusion: Impact of a Top Retailer Leaving a Key Region**  

This analysis highlights potential risk if foot locker leaves southeast region, which is the top-performing retailer. As we can see Foot Locker contributes 59.6M to the Southeast region's total sales of 163.1M, which indicates that foot locker is critical revenue driver for Adidas.
